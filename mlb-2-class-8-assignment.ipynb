```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('/kaggle/input/datasets/lovishbansal123/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
```

# Visualize at least 1 graph of each category. You can take help from the reference notebook

# Univariate Analysis

# 1. Pie chart

```python
data=titanic['Sex'].value_counts().sort_index()
plt.pie(data,
        labels=['male','female'],
        autopct="%0.2f%%")
plt.show()
```

```python
data1=iris['species'].value_counts().sort_index()
plt.pie(data1,labels=['setosa','versicolor ','virginica'],
        colors=['skyblue','brown','green'],
        autopct="%0.2f%%")
plt.show()
```

# 2. Count plot

```python
sns.countplot(data=tips,x='size')
plt.title('Size VS Count')
plt.show()
```

# 3. Kde + Rug plot

```python
sns.kdeplot(data = titanic, x = 'Fare')
sns.rugplot(data = titanic, x = 'Fare')
```

# 4. Violin plot

```python
sns.violinplot(data = iris['petal_length'],color='skyblue')
plt.title('ViolinPlot on Petal Length')
plt.show()
```

# 5. Box plot

```python
sns.boxplot(data = iris['petal_length'],color='green')
plt.title('BoxPlot on Petal Length')
plt.show()
```

# 6. Boxen plot

```python
sns.boxenplot(data = titanic['Fare'],color='green')
plt.title('BoxenPlot on Petal Length')
plt.show()
```

# 7. Histogram

```python
sns.histplot(data = iris['petal_length'],color='red')
plt.title('Histogram on Petal Length')
plt.show()
```

# Bivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip',)
plt.title('Scatterplot on tips')
plt.show()
```

# 2. Line plot

```python
sns.lineplot(data = tips, x = 'total_bill', y = 'tip',)
plt.title('Lineplot on tips')
plt.show()
```

# 3. Joint plot

```python
sns.jointplot(data = tips, x = 'total_bill', y = 'size',)
plt.title('Jointplot on tips')
plt.show()
```

# 4. Hexbin plot

```python
plt.hexbin(tips['total_bill'], tips['tip'],gridsize=30, cmap='Blues')
plt.colorbar(label='count in bin')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Hexbinplot on tips')
plt.show()
```

# 5. 2D kde plot

```python
sns.kdeplot(tips,x='total_bill', y='tip',fill=True)
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('2D kdeplot on tips')
plt.show()
```

# 6. Box plot

```python
sns.boxplot(tips,x='sex', y='smoker',palette = 'Set3')
plt.title('Boxplot on tips')
plt.show()
```

# 7. Violin plot

```python
sns.violinplot(tips,x='sex', y='smoker',palette = 'Set1')
plt.title('Violinplot on tips')
plt.show()
```

# 8. Swarm plot

```python
sns.swarmplot(data = tips, x='total_bill',y='day',hue='day')
plt.title('Swarm plot between total bill and day of Tips dataset')
plt.show()
```

# 9. Count plot

```python
sns.countplot(data = tips, x='size',hue='day')
plt.title('Count plot between size and day of Tips dataset')
plt.show()
```

# Multivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(data = tips, x = 'total_bill', y = 'tip',hue='day')
plt.title('Scatterplot on tips')
plt.show()
```

# 2. Pair plot

```python
sns.pairplot(tips, hue = 'day')
plt.show()
```

# 3. Heatmap

```python
sns.heatmap(titanic.corr(numeric_only = True))
plt.show()
```

# 4. FacetGrid plot

```python
g = sns.FacetGrid(iris,col = 'species')
g.map_dataframe(sns.histplot,x = 'petal_length')
plt.show()
```

# 5. PairGrid

```python
g = sns.PairGrid(tips)
g.map(sns.lineplot)
plt.show()
```

# 6. Joint plot

```python
sns.jointplot(data = titanic, x = 'Pclass', y = 'Survived', hue = 'Age')
plt.title('Multivariable Jointplot')
plt.show()
```

# 7. Boxen plot

```python
sns.boxenplot(titanic, x='Pclass', y='Age', hue='Survived')
plt.title('Boxen plot on Titanic')
plt.show()
```
