# Merging and joining

```python
students = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                   111, 112, 113, 114, 115, 116, 117, 118, 119, 120],
    "name": ["Rahim", "Karim", "Hasan", "Sakib", "Nabil",
             "Tanvir", "Fahim", "Rafi", "Imran", "Shakil",
             "Arif", "Jahid", "Mehedi", "Siam", "Tareq",
             "Rakib", "Adnan", "Nayeem", "Foysal", "Rony"],
    "department": ["CSE", "EEE", "CSE", "BBA", "CSE",
                   "EEE", "CSE", "BBA", "CSE", "EEE",
                   "CSE", "BBA", "EEE", "CSE", "BBA",
                   "CSE", "EEE", "CSE", "BBA", "EEE"]
})

students
```

```python
results = pd.DataFrame({
    "student_id": [101, 102, 103, 105, 106, 108, 109, 111, 112, 114,
                   121, 122, 123, 124, 125, 126, 127, 128, 129, 130],
    "math": [85, 72, 91, 68, 77, 88, 74, 95, 81, 69,
             78, 84, 92, 73, 89, 67, 80, 94, 76, 83],
    "python": [90, 75, 88, 72, 80, 92, 79, 96, 85, 71,
               82, 87, 95, 76, 91, 70, 84, 93, 78, 86]
})

results
```

```python
departments = pd.DataFrame({
    "dept_id": [1, 2, 3, 4],
    "department": [
        "CSE",
        "EEE",
        "BBA",
        "Civil"
    ],
    "faculty": [
        "Engineering",
        "Engineering",
        "Business",
        "Engineering"
    ]
})

departments
```

# 1. Create a dataframe contains students name,faculty and number on math and python

```python
data = students.merge(departments, how='inner', on='department')
data = data.merge(results, how='inner', on='student_id')
data = data[['name', 'faculty', 'math', 'python']]
data
```

# 2. Department wise students count

```python
students.groupby('department')['student_id'].count()
```

# 3. Faculty wise department count

```python
departments.groupby('faculty')['department'].count()
```

# 4. Find department wise highest number on each subject

```python
t = students.merge(results, on='student_id')
t.groupby('department')[['math', 'python']].max()
```

# 5. Find faculty wise highest number on each subject

```python
data.groupby('faculty')[['math', 'python']].max()
```
